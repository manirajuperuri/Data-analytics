```python
length=float(input("Enter the length of the rectangle: "))
breadth=1/2
area=length*breadth
perimeter=2*(length+breadth)
print("Area is:",area)
print("perimeter is:",perimeter)
```

```python
side=float(input("Enter the side of the square: "))
area=side*side
perimeter=4*side
print("Area is:",area)
print("perimeter is:",perimeter)
```

```python
circle=float(input("Enter the radius of the circle: "))
area=3.14*circle*circle
perimeter=2*3.14*circle
print("Area is:",area)
print("perimeter is:",perimeter)
```

```python
tri=float(input("Enter the base of the triangle: "))
height=float(input("Enter the height of the triangle: "))
area=1/2*tri*height
perimeter=tri+height+((tri**2)+(height**2))**0.5
print("Area is:",area)
print("perimeter is:",perimeter)
```

```python
#Shopping Bill Calculation Calculate the total bill amount for purchased products. Inputs: product_price, quantity

product_price=float(input("Enter the product price: "))
quantity=int(input("Enter the quantity: "))
product2_price=float(input("Enter the product price: "))
quantity2=int(input("Enter the quantity: "))
product3_price=float(input("Enter the product price: "))
quantity3=int(input("Enter the quantity: "))
total_bill=(product_price*quantity) + (product2_price*quantity2) + (product3_price*quantity3)
print("Total bill amount is:",total_bill)
```

```python
#ATM Withdrawal Calculate the remaining bank balance after a withdrawal. Inputs: account_balance, withdrawal_amount
account_balance=float(input("Enter the account balance: "))
withdrawal_amount=float(input("Enter the withdrawal amount: "))
remaining_balance=account_balance-withdrawal_amount
print("Remaining bank balance is:",remaining_balance)
```

```python
#. Student Marks Average Calculate the average marks obtained in five subjects. Inputs: subject1, subject2, subject3, subject4, subject5
subject1=float(input("Enter the marks in subject1: "))
subject2=float(input("Enter the marks in subject2: "))
subject3=float(input("Enter the marks in subject3: "))
subject4=float(input("Enter the marks in subject4: "))
subject5=float(input("Enter the marks in subject5: "))
average_marks=(subject1+subject2+subject3+subject4+subject5)/5
print("Average marks obtained is:",average_marks)
```
